```html
<!DOCTYPE html>
<html lang="id">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Untuk Dewi Ku Sayang</title>
    <style>
        body {
            display: flex;
            flex-direction: column;
            justify-content: center;
            align-items: center;
            min-height: 100vh;
            background-color: #fce4ec;
            font-family: 'Arial', sans-serif;
            color: #d81b60;
            margin: 0;
            overflow: hidden; /* Prevent scrollbars if buttons move off-screen slightly */
        }
        .container {
            background-color: #fff;
            padding: 40px 60px;
            border-radius: 15px;
            box-shadow: 0 10px 30px rgba(0, 0, 0, 0.1);
            text-align: center;
            max-width: 90%;
        }
        h1 {
            font-size: 2.5em;
            margin-bottom: 25px;
            color: #e91e63;
        }
        p {
            font-size: 1.2em;
            margin-bottom: 30px;
            line-height: 1.6;
        }
        .buttons {
            position: relative;
            height: 80px; /* Space for buttons to move */
            display: flex;
            justify-content: center;
            gap: 20px;
        }
        .btn {
            padding: 15px 30px;
            font-size: 1.1em;
            border: none;
            border-radius: 30px;
            cursor: pointer;
            transition: all 0.3s ease;
            font-weight: bold;
            text-decoration: none;
            display: inline-block;
        }
        #yesBtn {
            background-color: #4CAF50;
            color: white;
        }
        #yesBtn:hover {
            background-color: #45a049;
            transform: scale(1.05);
        }
        #noBtn {
            background-color: #f44336;
            color: white;
            position: absolute;
            top: 0;
            left: 50%;
            transform: translateX(50%); /* Center it initially relative to its container */
        }
        #noBtn:hover {
            background-color: #da190b;
        }
        .result-message {
            margin-top: 30px;
            font-size: 1.5em;
            font-weight: bold;
            color: #2e7d32;
        }
    </style>
</head>
<body>
    <div class="container">
        <h1>Halo Dewi Ku Sayang!</h1>
        <p>Dewi, ada sesuatu yang ingin kutanyakan padamu. Maukah kamu menjadi pacarku?</p>
        <div class="buttons">
            <button id="yesBtn" class="btn">Ya</button>
            <button id="noBtn" class="btn">Tidak</button>
        </div>
        <div id="result" class="result-message" style="display:none;"></div>
    </div>

    <script>
        const yesBtn = document.getElementById('yesBtn');
        const noBtn = document.getElementById('noBtn');
        const resultDiv = document.getElementById('result');
        const container = document.querySelector('.container');
        const buttonsContainer = document.querySelector('.buttons');

        yesBtn.addEventListener('click', () => {
            resultDiv.style.display = 'block';
            resultDiv.style.color = '#2e7d32';
            resultDiv.innerHTML = 'YEAY! Aku sangat bahagia! Aku cinta kamu, Dewi! ❤️';
            noBtn.style.display = 'none'; // Hide no button after 'Yes'
            yesBtn.style.display = 'none'; // Hide yes button after 'Yes'
        });

        noBtn.addEventListener('mouseover', () => {
            // Make the button jump to a random position within the window
            const viewportWidth = window.innerWidth;
            const viewportHeight = window.innerHeight;

            const buttonWidth = noBtn.offsetWidth;
            const buttonHeight = noBtn.offsetHeight;

            const newX = Math.random() * (viewportWidth - buttonWidth);
            const newY = Math.random() * (viewportHeight - buttonHeight);

            noBtn.style.position = 'fixed'; // Use fixed to position relative to viewport
            noBtn.style.left = `${newX}px`;
            noBtn.style.top = `${newY}px`;
            noBtn.style.transform = 'none'; // Remove initial transform for proper positioning
        });

        noBtn.addEventListener('click', (e) => {
            // Prevent default click behavior to ensure mouseover handles movement
            e.preventDefault();
            // Trigger mouseover logic directly on click as well
            noBtn.dispatchEvent(new Event('mouseover'));
        });

        // Reset button position if window resized and mouse is not over it
        window.addEventListener('resize', () => {
            // If the noBtn is currently fixed, reset its position
            if (noBtn.style.position === 'fixed') {
                // Optionally, recenter or move it again if it went off screen
                const viewportWidth = window.innerWidth;
                const viewportHeight = window.innerHeight;

                const buttonWidth = noBtn.offsetWidth;
                const buttonHeight = noBtn.offsetHeight;

                const newX = Math.random() * (viewportWidth - buttonWidth);
                const newY = Math.random() * (viewportHeight - buttonHeight);

                noBtn.style.left = `${newX}px`;
                noBtn.style.top = `${newY}px`;
            }
        });

    </script>
</body>
</html>
```
